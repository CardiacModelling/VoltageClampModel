## Some ideas

Below are three sketches of a patch, and ideas about where leak may arise.

<img src="./img/leak-extended.png" />

On the left, the pipette (and other electrode components) contribute a resistance $R_e$ or equivalently a conductance $g_e$, which arises mostly in the narrow tip.
Access to the cell contributes another resistance $R_a$ or equivalently $g_a$.
When the seal was made, membrane was sucked into and around the pipette, so that any contact with the bath would be at a point between $a$ and $e$, which we've labelled $V_1$.
The leak is indicated with an orange arrow from this point to the bath.

In the middle sketch the seal is much looser, and part of the "seal" is made up of debris.
In this scenario, there are two leak currents: one from the pipette tip through the debris into the bath, and one from the inside of the cell to the bath.
In this case, we could also imagine the debris to have a capacitance.
We could also draw a capacitance for the connection from the cell to the bath, but this would be indistinguishable from a slightly larger $C_m$, so this is left out.

On the right, the pipette and cell still contribute $g_e$ and $g_a$, but now, perhaps because of debris again, the leak current is from the cell interior directly to the bath.

## Equations 1: No $C_1$

Because the left and right schematics are a special case of the centre panel, we'll only treat the more complicated schematic.
To start, we'll leave out the capacitance $C_1$.

The total leak current is given by
\begin{align}
I_L = g_1 V_1 + g_2 V_m
\end{align}

The sum of currents at the $V_m$ node is
\begin{align}
g_a (V_1 - V_m) &= g_2 V_m + C_m \dot{V}_m + I \\
V_1 &= \frac{g_a + g_2}{g_a} V_m + \frac{I + C_m \dot{V}_m}{g_a}
\end{align}
where $I$ is the current of interest.
Combining, we find
\begin{align}
I_L &= \frac{g_1(g_a + g_2)}{g_a} V_m + \frac{g_1}{g_a}(I + C_m \dot{V}_m) + g_2 V_m \\
    &= \left(g_1 + g_2 + \frac{g_1 g_2}{g_a}\right) V_m + \frac{g_1}{g_a}(I + C_m \dot{V}_m)
\end{align}

For the **left schematic** we set $g_2 = 0$ to find
\begin{align}
I_L^\text{left} &= g_1 V_m + \frac{g_1}{g_a}(I + C_m \dot{V}_m)
\end{align}

For the **centre schematic** we can lump the unknown g's together to find
\begin{align}
I_L^\text{center} = g_x V_m + g_y (I + C_m \dot{V}_m) \\
\end{align}

And for the **right schematic** we set $g_1 = 0$ for
\begin{align}
I_L^\text{right} = g_1 V_m
\end{align}

### Interpretation

- None of these forms introduce a non-zero reversal potential $E_\text{leak}$.
- The added term $g I$ would be indistinguishable from a slightly larger current-of-interest, so would be absorbed by the model fit.
- So in this model only $g \, C_m \dot{V}_m$ is a new _and observable_ term.

For a very small current $I$, the derivative $\dot{V}_m$ would be dominated by membrane charging and discharging, and so this new leak current would appear as positive and negative decaying exponentials after every step change.
This is not very similar to what we see.

But for larger $I$, would the derivative $\dot{V}_m$ take on some features of $I$?
