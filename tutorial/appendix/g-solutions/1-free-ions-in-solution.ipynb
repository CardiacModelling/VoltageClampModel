# Bath & pipette solutions

In this appendix we look at bath and pipette solutions _from a modelling point of view_, and

- Determine **_free_** ion concentrations from the listed _added_ concentrations.
- Use these to estimate the **liquid junction potential (LJP)** and the **reversal (Nernst) potential** for potassium (EK).
- Estimate **errors** in the concentrations, and propagate these to LJP and EK.

We assume some mathematical knowledge, but very limited practical and theoretical knowledge of (bio)chemistry.

In this notebook we describe and analyse the general method of preparing solutions.
The next two notebooks give real-world examples.

## General method of preparing solutions

Bath and pipette solutions are prepared by mixing powders (e.g. of [KCl](https://www.sigmaaldrich.com/GB/en/product/sigma/p9541)) and stock solutions (e.g. of [$\text{CaCl}_2$](https://www.sigmaaldrich.com/GB/en/product/sigma/21115)), adding purified water until the desired volume is reached, and adjusting the pH to e.g. 7.4 (external) or 7.2 (internal) by adding a few drops of a basic or acidic solution.
As the pH adjustment slightly increases the total volume, a more accurate method is to

1. Combine the ingredients
2. Add pure water till just below the desired volume, and stir
3. Adjust the pH
4. Set the final volume

### Adding by weigth and by volume


For example, to make $100\,\text{mL}$ solution with target concentration $10\,\text{mM}$ KCl, we would look up the molecular weight of KCl to find $74.55\,\text{g}/\text{mol}$, and then measure out
\begin{align}
0.1\,\text{L} \cdot 10\,\frac{\text{mmol}}{\text{L}} \cdot 74.55\,\frac{\text{g}}{\text{mol}} = 74.55\,\text{mg}
\end{align}
using very accurate scales.


Similarly, if we want 2 mM $\text{CaCl}_2$ in the same mix, we look up the concentration (e.g. $1\,\text{M}$) and use a pipette to add 
\begin{align}
0.1\,\text{L} \cdot 2\,\frac{\text{mmol}}{\text{L}} / 1\,\frac{\text{mol}}{\text{L}} = 0.2\,\text{mL}
\end{align}
of the stock solution.




## Purity

For compounds added by weight, the supplier usually lists a _purity_.
This specifies how much of the powder is actually the compound of interest.
Although given as a lower bound, it is common practice to interpret it as the true value, and used to adjust the added weight according to:
\begin{align}
m = \frac{C_t M V}{p} 
&\quad&
\text{g} = \frac{\text{mmol}}{\text{L}} \frac{\text{g}}{\text{mol}} \text{L}
\end{align}
where $m$ is the weight required to obtain the target solution, and common units are shown on the right.

For compounds added by volume, purity is often not supplied, so we simply use
\begin{align}
V_a = \frac{C_t}{C_a} V
&\quad&
\text{mL} = \frac{\text{mmol}}{\text{L}} \cdot \frac{\text{L}}{\text{mol}} \cdot \text{L}
\end{align}
where $V_a$ is the volume to add, and common units are shown on the right.

### "Loss on drying"

A second factor affecting purity, listed for [EGTA](https://www.sigmaaldrich.com/GB/en/product/sigma/e3889) in the table above, is water content.
This occurs when $\text{H}_2\text{O}$ molecules are embedded in a crystaline structure, and is measured (but not exactly the same as) as "loss on drying".

This is given as an upper bound, for example $\leq 1\%$, but as with purity, it is commonly factored in by reinterpreting as the exact value:
\begin{align}
m = \frac{C_t M V}{p_1 (1 - p_2)} 
\end{align}
where $p_1$ is purity (e.g. 0.97 for $\geq 97\%$) and $p_2$ is loss on drying (e.g. 0.01 for $\leq 1\%$).


### Unspecified water content

Sometimes, water content is explicitly acknowledged in the given formula.
For example, for this [ATP disodium salt hydrate](https://www.sigmaaldrich.com/GB/en/product/aldrich/a26209) the formula is given as

\end{align}
where the $A \cdot x B$ notation means "a crystal of A with embedded B, at a ratio of x B per A", see for example the wikipedia page on [hydrates](https://en.wikipedia.org/wiki/Hydrate) and [water of crystallisation](https://en.wikipedia.org/wiki/Water_of_crystallization).

\end{align}
so that we can see that two Na have took over for two H.

But what is the value of $x$?
Unfortunately, the page specifies

As an estimate, we can look at [a different supplier](https://www.thermofisher.com/order/catalog/product/L14522.06), who list the water content as $\leq 10\%$.

where $p_1$ is purity (0.98 for this product) and $p_2$ is water content (the estimated 0.1).

### Unspecified magnesium content

Another unspecified ratio occurs with the "$\text{Mg}^{2+}\text{ATP}$" listed in our example above.
On [the supplier website](https://www.sigmaaldrich.com/GB/en/product/sigma/a9187) the formula is given as
\begin{align}
\text{C}_{10}\text{H}_{16}\text{N}_5\text{O}_{13}\text{P}_3 \cdot x\text{Mg}^{2+}
\end{align}
which shows that this is an ATP crystal embedding magnesium.
Near the bottom of the page, the ratio of moles of Mg to moles of ATP is specified as approximately 1 to 1.5.

But even if we knew $x$, because we are interested in both Mg and ATP, we cannot adjust the weight to correct for both simultaneously.
Instead, we will have to treat this as a source of _uncertainty_.
And because the molecular weight of the crystal is dominated by the weight of ATP (even for quite a large $x$), this will mostly be uncertainty in $[\text{Mg}^{2+}]$.


## Adjusting the pH

Once we have worked out how much to add, we mix everything together and then measure the pH.
Solutions used in cardiac cell electrophysiology are typically acidic before adjustment, for example due to the presence of HEPES, EGTA, or ATP in powder form, will shed a lot of protons ($H^+$) when dissolved.
As a result, the pH needs to be adjusted by adding a strong base such as NaOH of KOH.
**This can add a substantial amount of Na or K, particularly for internal solutions, and so will affect the LJP and reversal potentials.**

When setting up new experiments, if the exact concentrations are important it is a good idea to use a measuring pipette to add the base one drop at a time, count the drops, and calculate the added Na or K.

When analysing existing data, this amount is almost never reported.
Instead, it may be possible to estimate the added concentration by assuming the total charge in the solution is zero (see Example 1).

### The pH, $[\text{H}^+]$, $[\text{OH}^-]$, and the LJP

Since our goal is to calculate the concentrations of free ions, do we need to account for released $\text{H}^+$ and added $\text{OH}^-$ when calculating the LJP?

Some background information will help us work this out:
Even before adding anything, pure water is mostly $\text{H}_2\text{O}$, but a small fraction [self-ionises](https://en.wikipedia.org/wiki/Self-ionization_of_water) to become $\text{H}_3\text{O}^+$ and $\text{OH}^-$.
The notation "$\text{H}^+$" is often used as a shorthand for "$\text{H}_3\text{O}^+$", but in reality any free $\text{H}^+$ almost instantly binds to a nearby $\text{H}_2\text{O}$.
The pH is a measure of the $\text{H}_3\text{O}^+$ concentration, and is defined as:
\begin{align}
\text{pH} = -\log_{10}(a_{\text{H}^+}) \approx -\log_{10}\left([\text{H}^+]\right)
\end{align}
where $a_{\text{H}^+}$ is the activity of $\text{H}_3\text{O}^+$, which is approximated by its molar concentration.
A similar measure, the pOH is approximated by $-\log_{10}\left([\text{OH}^-]\right)$, and at room temperature the two are related by $\text{pOH} \approx 14 - \text{pH}$.

For a pH of 7.4 this gives $[\text{H}^+] \approx 4 \cdot 10^{-8}$ M, or $4 \cdot 10^{-5}$ mM.
For pH 7.2 we find $[\text{H}^+] \approx 6 \cdot 10^{-5}$ mM.
Using $\text{pOH} = 14 - \text{pH}$ we can see that $[\text{OH}^-]$ will be similarly small.

**As a result, $[\text{H}^+]$ and $[\text{OH}^-]$ can be neglected in LJP calculations at physiological pH**.
Similarly, they can be omitted when calculating the total charge of the solution, or its "ionic equivalence" (see below).

## From target concentration to free concentrations

So far, we have concentrated on the target concentrations, and working out the weight or volume to add to the mix to achieve them.
The next question is what happens once all the ingredients dissolve and mix.
Salts like NaCl or KCl dissolve practically completely in water, and since there are typically no substances binding Na, K, or Cl in our solutions, we can assume that adding 10mM NaCl adds 10mM free $[\text{Na}^+]$.

But for other compounds, we need to deal with further complications, including:

1. Buffers of the pH such as HEPES, which will exist in a protonated fraction ($[\text{HEPES}]$) and a deprotonated fraction ($[\text{HEPES}^-]$) which depends on the pH.
2. Other compounds with multiple protonation states, such as EGTA, which is added as uncharged $\text{H}_4\text{EGTA}$, but can lose up to four protons to become $\text{EGTA}^{4-}$. At physiological pH, it exists almost exclusively as $\text{H}_2\text{EGTA}^{2-}$.
3. Compounds that bind to ions, for example ["chelation"](https://en.wikipedia.org/wiki/Chelation) of $\text{Ca}^{2+}$ by EGTA, and of $\text{Mg}^{2+}$ by ATP.

## Buffering with HEPES

HEPES can exist as either $[\text{HEPES}]$ or $[\text{HEPES}^-]$, with the exact ratio between the two depending on the pH.
This is described by the [Henderson-Hasselbalch equation](https://en.wikipedia.org/wiki/Henderson%E2%80%93Hasselbalch_equation):
\begin{align}
\text{pH} = \text{pK}_a + \log_{10}\left( \frac{[\text{HEPES}^-]}{[\text{HEPES}]} \right)
\end{align}
Where $\text{pK}_a = -\log_{10} K_a$ is HEPES's [acid dissociation rate constant](https://en.wikipedia.org/wiki/Acid_dissociation_constant).

The "a" in "pKa" stands for "acid", which can be confusing, as the notation $K_a$ is also used to mean a general _association_ constant (and so equal to $1/K_a$ in the "a for acid" notation).

We can combine the equation above with
\begin{align}
[\text{HEPES}^-] + [\text{HEPES}] = [\text{HEPES}]_\text{added}
\end{align}
to find
\begin{align}
[\text{HEPES}] = \frac{[\text{HEPES}]_\text{added}}{1 + 10^{\text{pH} - \text{pK}_a}}
&&\quad&&
[\text{HEPES}^-] = \frac{[\text{HEPES}]_\text{added}}{1 + 10^{\text{pK}_a - \text{pH}}}
\end{align}

A value for HEPES' pKA can be found e.g. [on this website](https://condor.depaul.edu/jmaresh/341/Lab1/Lab_1_Buffers4.html), which lists it as 7.55 at $20\,^\circ\text{C}$, with a temperature dependent shift of -0.014 per degree celsius.
So, for example, if the pH of a solution with 10mM HEPES was set to 7.4 at $22\,^\circ\text{C}$, the pKa was 7.522, yielding
\begin{align}
[\text{HEPES}] = 5.7\,\text{mM}
&&\quad&&
[\text{HEPES}^-] = 4.3\,\text{mM}
\end{align}

### Temperature change

What happens if the temperature is changed _after the pH was set_?

Here, [the same website](https://condor.depaul.edu/jmaresh/341/Lab1/Lab_1_Buffers4.html) says something confusing:

> ... if you calculate the change in pKa with temperature, the pH of the buffer will change the same amount because changing the temperature will not change the ratio of the basic form to acid form ... raising the temperature of HEPES by 1.0 °C lowers both the pKa and pH by 0.014.

In the light of the equation above, this sounds impossible: for the pH to change, HEPES must lose or gain protons, and so the protonation ratio _must_ change.
However, changing the pH from e.g. 7.2 to 7.1 implies a change from $6.3 \cdot 10^{-5}\,\text{mM}$ free $\text{H}^+$ to $7.9 \cdot 10^-5\,\text{mM}$, so that the change in HEPES protonation ratio is _practically_ the same.

So if we set the pH at 22 degrees, but measure at $37\,^\circ\text{C}$, the ratio of charged to uncharged HEPES will be roughly the same.
The pH, however, will have changed.
**For this reason, it is always good to either set the pH at the temperature the experiments will be run**, or alternatively to calculate the expected shift and adjust for this accordingly.

## $\text{Ca}^{2+}$, EGTA, $\text{Mg}^{2+}$, and ATP

The free calcium concentration inside a cell is both very low, and tightly controlled.
To recreate this situation in pipette solution, it is common to add a very strong calcium buffer, such as EGTA, and to choose the buffer-to-calcium ratio to achieve a specific micromolar free calcium concentration.

ATP, the main source of energe inside cells, requires magnesium to be active (see e.g. [wikipedia](https://en.wikipedia.org/wiki/Magnesium_in_biology)), and is present mostly as $\text{MgATP}^{2-}$ inside cells.
To recreate this in pipette solution, we either add both seperately (e.g. with $\text{MgCl}_2$ and [ATP disodium salt](https://www.sigmaaldrich.com/GB/en/product/aldrich/a26209)) or already together (e.g. with [ATP magnesium salt hydrate](https://www.sigmaaldrich.com/GB/en/product/sigma/a9187)).

In addition, EGTA can bind magnesium, and ATP can bind calcium.
And so, **if accurate estimates of free calcium are required**, all four need to be considered simultaneously.
However, as we show in Example 1, **approximate methods ignoring Mg-EGTA and Ca-ATP binding are enough to determine the LJP**.

### Considering all four at once

To calculate free and bound concentrations of $\text{Ca}^{2+}$, $\text{Mg}^{2+}$, Ca, EGTA and ATP, we can use online calculators.
Most famously, [Maxchelator](https://somapp.ucdmc.ucdavis.edu/pharmacology/bers/maxchelator/downloads.htm), described in [Bers, Patton & Nuccitelli 2010](https://doi.org/10.1016/S0091-679X(08)61108-5), is still available online although backing up the HTML (containing hardcoded JS) is recommended.
A more recent effort is [PyChelator](https://amrutelab.github.io/PyChelator/), which is available as a Colab notebook or in Javascript, and described in [Spahiu, Kastrati & Amrute-Nayak 2024](https://doi.org/10.1186/s12859-024-05858-8), with code on [GitHub](https://github.com/AmruteLab/PyChelator).

Both calculators take the total $\text{Ca}^{2+}$, $\text{Mg}^{2+}$, EGTA, and ATP as input, and then calculate bound and free fractions for each.
(Unfortunately, they do not show what has bound to what, but see below).

To do the calculations, the programs
1. Load a set of binding constants
2. Adjust these, based on the **pH, temperature, and "ionic equivalence"**
3. Iteratively work out an equilibrium

The procedure for adjusting the constants is described in the paper by Bers et al. 
The value for "ionic equivalence" is found as
\begin{align}
I_e = \frac{1}{2} \sum{C_i |z_i|}
\end{align}
where $C_i$ is the concentration of an ionic species, $|z_i|$ is the magnitude of its valence, and the sum is over all species in the solution.
For the choice of ionic equivalence over the more conventional ionic strength ($\frac{1}{2} \sum{C_i z_i^2}$), see [Smith & Miller, 1984](https://doi.org/10.1016/0304-4165(85)90011-X).

This presents us with a problem: We need to know the ionic concentrations in order to calculate them.
Luckily, the influence of $I_e$ is not particularly strong, and so we can use a rough estimate in the calculations, for example based on the methods described below.

### Simplification: Calcium and EGTA in isolation

EGTA is a strong and highly specific calcium buffer.
In a solution with more EGTA than calcium, practically all calcium will be bound, and any change to the _total_ amount of calcium, $[\text{Ca}]_\text{total} = [\text{Ca}^{2+}]_\text{free} + [\text{Ca}^{2-}]_\text{bound}$, will lead to a much smaller change in the amount of _free_ calcium.

Both free EGTA and CaEGTA can exist in multiple protonation states (see [Dweck, Reyes-Alfonso Jr., and Potter](https://doi.org/10.1016/j.ab.2005.09.025)), but here we make the following simplifying assumptions (for physiological pH and room or body temperature):

1. All free EGTA is present as $\text{H}_2\text{EGTA}^{2-}$
2. All bound EGTA is present as $\text{Ca}\text{EGTA}^{2-}$
3. All EGTA and $\text{Ca}^{2+}$ that can bind, will bind.

Under these assumptions, a solution with target concentrations of $1\,\text{mM}\,[\text{Ca}^{2+}]$ and $10\,\text{mM}\,\text{EGTA}$ will result in free concentrations $[\text{Ca}^{2+}]_\text{free} = 0$, $[\text{CaEGTA}^{2-}] = 1\,\text{mM}$, and $[\text{H}_2\text{EGTA}^{2-}] = 9\,\text{mM}$.


### Simplification: Magnesium and ATP in isolation

As with calcium and EGTA, we can make the simplifying assumption that

1. All $\text{Mg}^{2+}$ and ATP that can bind, will bind

So in a solution with target concentrations $[\text{Mg}^{2+}]= 6\,\text{mM}$ and $[\text{ATP}]= 5\,\text{mM}$ we will have $[\text{MgATP}^{2-}]=5\,\text{mM}$, $[\text{Mg}^{2+}]=1\,\text{mM}$ and $[\text{ATP}]_\text{free}=0\,\text{mM}$.

If there is more ATP than magnesium, we have a non-zero concentration of free ATP, and we need to work out the protonation state of the free ATP.
At physiological pH, there is a mixture of $\text{ATP}^{4-}$ and $\text{HATP}^{3-}$, with a pKa of approximately 6.5 (see [Stockbridge & Wolfenden](https://doi.org/10.1074/jbc.M109.017806)).
Using the same equations as for HEPES, we find
\begin{align}
[\text{ATP}^{4-}] = \frac{[\text{ATP}]_\text{free}}{1 + 10^{\text{pK}_a - \text{pH}}}
&&\quad&&
[\text{HATP}^{3-}] = \frac{[\text{ATP}]_\text{free}}{1 + 10^{\text{pH} - \text{pK}_a}}
\end{align}
where
\begin{align}
[\text{ATP}]_\text{free} = [\text{ATP}^{4-}] + [\text{HATP}^{3-}]
\end{align}

### Improving on the simplifications

If we do want to know the exact concentrations, we

1. Use the simplifications above to obtain a provisional result
2. Calculate the ionic equivalence in the provisional result
3. Use maxchelator or pychelator to obtain a more accurate result

Unfortunately neither software provides the exact concentrations of bound compounds, and will instead provide a table listing total and bound concentrations of $\text{Ca}^{2+}$, $\text{Mg}^{2+}$, EGTA and ATP individually.
In example 1 we show how similar simplifications as employed above can be used to work out the concentrations of $[\text{CaEGTA}^{2-}]$, $[\text{MgEGTA}^{2-}]$, etc.

## Calculating the LJP and reversal potentials

Using the above, we can make a table of all free ionic species in the pipette and bath solutions.
These can then be fed into LJP calculators such as [LJPCalc](https://swharden.com/LJPcalc/).

Reversal potentials for single-ion currents are calculated from the [Nernst equation](https://en.wikipedia.org/wiki/Nernst_equation):
\begin{align}
E = \frac{RT}{F} \log \frac{[\text{X}]_e}{[\text{X}]_i}
\end{align}
where the subscripts $e$ and $i$ denote external (bath) and internal (pipette).

For multi-ion currents, the [Goldman–Hodgkin–Katz voltage equation](https://en.wikipedia.org/wiki/Goldman_equation) can be used, but note that, for the canonical example - the L-type calcium current - we do need an accurate estimate of the free calcium concentration.

## Estimating the error in concentrations

Next, we estimate the error in target concentrations due to uncertainty in total volume, added weight or volume, and purity of the added compounds.
To do this, we use the method outlined below.

### Propagation of uncertainty

Given any function $f$ of a measurement $x \pm \Delta x$, we can estimate the error in $f(x)$ by taking a partial derivative and linearising:
\begin{align}
f(x) \rightarrow \Delta f = \left| \frac{\partial f}{\partial x} \right| \Delta x
\end{align}
For multiple _independent_ measurements, we can extend this to e.g.
\begin{align}
f(x, y, z) \rightarrow \Delta f = \sqrt{
    \left( \frac{\partial f}{\partial x} \Delta x \right)^2 +
    \left( \frac{\partial f}{\partial y} \Delta y \right)^2 +
    \left( \frac{\partial f}{\partial z} \Delta z \right)^2}
\end{align}
See e.g. [wikipedia on "propagation of uncertainty"](https://en.wikipedia.org/wiki/Propagation_of_uncertainty#Non-linear_combinations).
For intuition, we can use probability theory to interpret $\Delta f$ as a variance, or think geometrically and interpret it as a distance.

What this means for us is that we can write the target concentration as a function of the added weight/volume, the final volume, and purity, and work out how their measurement accuracy affects the final concentration.

### Adding substances by weight

To estimate the error when adding substances by weight, we define the target concentration $C_t$, the total volume $V$, the added mass $m$ in grams, the molecular weight $M$, and the purity $p$.
Then,
\begin{align}
C_t = \frac{m \cdot p}{M \cdot V} && \quad && \frac{\text{mmol}}{\text{L}} = \frac{\text{mg}}{\text{g}/\text{mol} \cdot \text{L}}
\end{align}

Assumimg that the molecular weight is accurate, the terms in our uncertainty equation are
\begin{align}
\frac{\partial C_t}{\partial m} \Delta m &= \frac{f}{M \cdot V} \Delta m = C_t\frac{\Delta m}{m} \\
\frac{\partial C_t}{\partial f} \Delta f &= \frac{m}{M \cdot V} \Delta f = C_t\frac{\Delta f}{f} \\
\frac{\partial C_t}{\partial V} \Delta V &= \frac{-m \cdot f}{M \cdot V^2} \Delta V = - C_t \frac{\Delta V}{V}
\end{align}
which we combine as
\begin{align}
\Delta C_t &= C_t\sqrt{\left(\frac{\Delta m}{m}\right)^2 + \left(\frac{\Delta f}{f}\right)^2 + \left(\frac{\Delta V}{V}\right)^2}
\end{align}

To calculate $m$, we use the target concentrations and the purities and molecular weights listed by a popular supplier.
\begin{align}
m = \frac{C_t \cdot M \cdot V}{f} && \quad && \text{mg} = \frac{\text{mmol}}{\text{L}} \frac{\text{g}}{\text{mol}} \text{L}
\end{align}
For the powders used in the external solution we find

| Compound | Ct     | M           | purity       | m (f=1) | m       |
|----------|--------|-------------|--------------|---------|---------|
| NaCl     | 140 mM | 58.44 g/mol | $\geq 99.0%$ | 8182 mg | 8264 mg |
| KCl      |   4 mM | 74.55 g/mol | $\geq 99.0%$ |  298 mg |  301 mg |
| HEPES    |  10 mM | 238.3 g/mol | $\geq 99.5%$ | 2383 mg | 2407 mg |

Here the last two columns list the weight to add when ignoring purity (f=1) or when using the stated lower bound value for f.
To estimate $\Delta m$ we will use both $1\,\text{mg}$ and $10\,\text{mg}$.

For $f$ and $\Delta f$ we will assume 99% purity means $99.5 \pm 0.5$ - leading to a slightly different $m$ than either column in the table above.

The volume was set using a volumetric flask, which is [highly accurate](https://medilabexports.com/class-a-vs-class-b-laboratory-glassware/), with a tolerance of either $\pm 0.3\,\text{mL}$ or $\pm 0.6\,\text{mL}$ for a 1L flask.

With this, we can calculate the expected errors in the sodium, potassium, and HEPES concentrations.
(For Cl, we could write an equation with two $m$ terms, but we can also approximate it with the error for Na.)

### Adding by volume

Similarly, when adding the volume $V_a$ of a stock solution with molar concentration $C_a$,
\begin{align}
C_t = \frac{C_a \cdot V_a}{V}
&& \quad &&
\frac{\text{mmol}}{\text{L}} = \frac{\text{mmol}}{\text{L}} \frac{\text{mL}}{\text{mL}}
\end{align}

Purity estimates for CaCl2 are not easy to find, so we will use
\begin{align}
\Delta C_t = C_t \sqrt{
    \left( \frac{\Delta V_a}{V_a} \right)^2 +
    \left( \frac{\Delta V}{V} \right)^2
}
\end{align}

### $\text{Ca}^{2+}$ and EGTA

EGTA is used as a calcium buffer: it very strongly binds $\text{Ca}^{2+}$, so that most of the calcium in a solution is in this bound form, and there is hardly any free calcium.
As a result, changes to the _total_ amount of calcium ($[\text{Ca}]_\text{total} = [\text{Ca}^{2+}]_\text{free} + [\text{CaEGTA}^{2-}]$) will lead to much smaller changes in the amount of _free_ calcium.

Calculating the free calcium concentration is complicated by (1) several protonation states of EGTA and CaEGTA and (2) interactions with $\text{Mg}^{2+}$ and ATP (see [Dweck, Reyes-Alfonso Jr., and Potter](https://doi.org/10.1016/j.ab.2005.09.025)).

To simplify, we will make the following assumptions:
1. All free EGTA is present as $\text{H}_2\text{EGTA}^{2-}$
2. All bound EGTA is present as $\text{Ca}\text{EGTA}^{2-}$
3. All EGTA and $\text{Ca}^{2+}$ that can bind, will bind.

Under these assumptions, in this solution we have no free calcium; 1 mM $\text{CaEGTA}^{2-}$; and 9 mM $\text{H}_2\text{EGTA}^{2-}$.

### $\text{Mg}^{2+}$ and ATP

Like calcium and EGTA, magnesium and ATP have a strong binding affinity, but now (1) the binding is not quite as strong, and (2) there is more magnesium than ATP in the solution.

Also like EGTA, ATP can exist in different protonation states.
At physiological pH, it is a mixture of $\text{ATP}^{4-}$ and $\text{HATP}^{3-}$, with a pKa of approximately 6.5 (see [Stockbridge & Wolfenden](https://doi.org/10.1074/jbc.M109.017806)).

\begin{align}
[\text{ATP}^{4-}] = \frac{[\text{ATP}]_\text{free}}{1 + 10^{\text{pK}_a - \text{pH}}}
&&\quad&&
[\text{HATP}^{3-}] = \frac{[\text{ATP}]_\text{free}}{1 + 10^{\text{pH} - \text{pK}_a}}
\end{align}
where
\begin{align}
[\text{ATP}]_\text{free} = [\text{ATP}^{4-}] + [\text{HATP}^{3-}]
\end{align}
for 83.4% $\text{ATP}^{4-}$ and 16.6% $\text{HATP}^{3-}$.

However, since there is more Mg than ATP in the solution, as a first approximation we will assume _all ATP is bound to Mg_, and $[\text{ATP}^{4-}] \approx [\text{HATP}^{3-}] \approx 0$.

## Osmolarity

This can be calculated (see e.g. [wikipedia](https://en.wikipedia.org/wiki/Osmotic_concentration)) from the target concentrations using

\begin{align}
\sum \phi_i n_i C_{t,i} \approx \sum n_i C_{t,i}
\end{align}

where $C_{t, i}$ is the target concentration for compound $i$ and $n_i$ is the number of particles the compound dissociates into (i.e. 1 for HEPES, glucose, and sucrose; 2 for NaCl and KCl; and 3 for CaCl2).
The [osmotic coefficient](https://en.wikipedia.org/wiki/Osmotic_coefficient) accounts for non-ideality of the dissociation, and will simply be set to 1 here.

With this, we can calculate the osmolarity of the external solution as
\begin{align}
(10 + 10) + 2 \cdot (130 + 5 + 36) + 3 \cdot (1 + 1)1.8 = 336.72
\end{align}

## Minimum conductance estimate

In Gary's group, when measuring IKr we like to start protocols with a step sequence that elicits a very strong current.
The reason for this is that when we want to estimate conductance $g$ and kinetic parameters $\theta$ in
\begin{align}
I = g \cdot O(\theta) (V - E)
\end{align}
then $g$ and $O(\theta)$ can compensate each other, making the system unidentifiable.
This can be remedied (somewhat) by having a step that brings $O$ very close to 1.

The size of the current in this step can then be used to set a _lower bound_ on $g$: the minimum conductance $g_\text{min}$:
\begin{align}
I_\text{big} \approx g_\text{min} \cdot 1 \cdot (V - E)
\end{align}



### Minimum conductance and reversal potential

A typical step used to measure $g_\text{min}$ is to -120mV.
Using the concentrations in this notebook, $E$ was calculated as $-99.25\,\text{mV}$, but only $-92.74\,\text{mV}$ when not taking titration into account.
At first glance, this could cause an underestimate of the minimum conductance of a factor
\begin{align}
\left(\frac{I_\text{big}}{-120 - -93}\right) / 
\left(\frac{I_\text{big}}{-120 - -99.25}\right)
= \frac{-20.75}{27} \approx 0.77
\end{align}

_However_, if we are already compensating for the difference between observed and calculated reversal potential, then both estimates are the same.



#### Compensating for fixed voltage offsets

**This was silly and can probably go**

If we suspect a fixed voltage offset (or do not want to model a time or voltage-dependent one), we can introduce a difference between applied and membrane voltage, as
\begin{align}
V_m = V_c + E_\text{off}
\end{align}
where $V_c$ is the intended (command) potential.

If our protocol includes a "reversal ramp" or another means of measuring the reversal potential, we can use the difference between the calculated (Nernst) and observed potential to set $E_\text{off}$.

Setting $V_c = E_\text{observed}$ and $V_m = E_\text{nernst}$, we find
\begin{align}
V_m             &= V_c + E_\text{off} \\
E_\text{nernst} &= E_\text{observed} + E_\text{off} \\
E_\text{off} &= E_\text{nernst} - E_\text{observed}
\end{align}

The driving force $D$ during a step to an intended $V_{c,\text{step}}$ (but an actual $V_{m,\text{step}}$) is then

\begin{align}
D &= V_\text{m, step} - E_\text{nernst} \\
  &= (V_\text{c, step} + E_\text{off}) - (E_\text{observed} + E_\text{off}) \\
  &= V_\text{c, step} - E_\text{observed}
\end{align}

and so independent of our calculated $E_\text{nernst}$.
