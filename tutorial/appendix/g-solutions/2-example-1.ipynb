**This example** describes a set of solutions used in IKr experiments (and as the example in the previous notebook).

Accepting these results, we can calculate the reversal potential for potassium from
\begin{align}
E &= \frac{RT}{F} \log \frac{[\text{K}^+]_e}{[\text{K}^+]_i}
\end{align}

Using $37^\circ\text{C}$ and the values from the table, we find $E_K = -99.6\,\text{mV}$.

Ignoring the effect of titration, we would get $E_K(130\,\text{mM}) = -93.0\,\text{mV}$.

## Uncertainty in the target concentrations

Finally, we look at uncertainty in the target concentrations that can arise when measuring out ingredients.

### Adding by weight

To estimate the uncertainty in concentration for substances added by weight, we need to assign values to the terms in
\begin{align}
\Delta C_t &= C_t\sqrt{\left(\frac{\Delta m}{m}\right)^2 + \left(\frac{\Delta V}{V}\right)^2 + \left(\frac{\Delta p}{p}\right)^2}
\end{align}

In this experiment, the external volume was set using a [highly accurate](https://medilabexports.com/class-a-vs-class-b-laboratory-glassware/) 1 litre _volumetric flask_.
These have a tolerance of either $\pm 0.3\,\text{mL}$ or $\pm 0.6\,\text{mL}$ so we can set an upper bound for $\frac{\Delta V}{V}$ of at $6\cdot10^{-4}$.

For $\Delta m$ we can look at specifications of "analytical balances", for example [this one](https://www.mt.com/us/en/home/phased_out_products/Laboratory_Weighing_Solutions/Analytical/Excellence/XS_Analytical_Balance/XS64.html) used in Example 2.
It shows several measures, including readability (how many digits are on the display, regardless of accuracy), repeatability (the standard deviation obtained if you keep measuring the same thing), and linearity (if you measure objects over the full allowable range, what is the maximum deviation from the true weight).
To find an upper bound, we could use the largest listed number of $\pm0.2\,\text{mg}$.
But to allow for some user error, we will use $\pm1\,\text{mg}$.

For $m$ itself, we can repeat the calculations for added weight to find $m$ for each compound in the external solution:

| Compound | Ct     | M           | p     | m       |
|----------|--------|-------------|-------|---------|
| NaCl     | 140 mM | 58.44 g/mol | 0.99  | 8264 mg |
| KCl      |   4 mM | 74.55 g/mol | 0.99  |  301 mg |
| HEPES    |  10 mM | 238.3 g/mol | 0.995 | 2395 mg |

Finally, a value for $\Delta p$ is a bit tricky, as purity is specified as a lower bound.
Here, we will take the pessimistic view that $\Delta p = 1 - p$.

Filling in these values for NaCl, KCl and HEPES, we find:

````python
import math

# Ct, M, purity
ions = {
    'NaCl': (140, 58.44, 99),
    'KCl': (4, 74.55, 99),
    'HEPES': (10, 238.3, 99.5),
}

# Relative error in final volume
dVV = 6e-4

# Absolute error in weight
dm = 1

for k, (Ct, M, p) in ions.items():
    p = p / 100
    dp = 1 - p
    m = Ct * M * 1 / p
    print(f'{k} (Ct={Ct} mM, m={m:.0f} mg)')
    dcc = math.sqrt(dVV**2 + (dp / p)**2 + (dm / m)**2)
    print(f'  |dV/V|: {abs(dVV):1.6g}')
    print(f'  |dp/p|: {abs(dp/p):1.6g}')
    print(f'  |dm/m|: {abs(dm / m):1.6g}')
    print(f'  dCt/Ct: {dcc:1.6f}')
    print(f'     dCt: {dcc * Ct:1.6f} mM')
````

So in this experiment, the relative error in sodium, concentration, and HEPES concentration is

1. Small, being at most 1.1%
2. Dominated by $\frac{\Delta p}{p}$, not $\frac{\Delta m}{m}$ or $\frac{\Delta V}{V}$

For the absolute error, we see that the compound with the highest concentration has the largest error.

Since purity dominates here, what happens if we interpret it a different way?
For example, we could interpret a 99% purity product as being $99.5\pm0.5%$ pure, which would halve $\Delta p$.
Or, staying closest to general practice, we could ignore uncertainty in $p$ althogether, setting $\Delta p = 0$.
In this case, we find

````python
for k, (Ct, M, p) in ions.items():
    p = p / 100
    m = Ct * M * 1 / p
    print(f'{k} (Ct={Ct} mM, m={m:.0f} mg)')
    dcc = math.sqrt(dVV**2 + (dm / m)**2)
    print(f'  |dV/V|: {abs(dVV):1.6g}')
    print(f'  |dm/m|: {abs(dm / m):1.6g}')
    print(f'  dCt/Ct: {dcc:1.6f}')
    print(f'     dCt: {dcc * Ct:1.6f} mM')
````

so that weighing becomes the dominant factor.

#### Compounds added by volume

For CaCl2, which was added by volume, we need to assign values to
\begin{align}
\Delta C_t = C_t \sqrt{
    \left( \frac{\Delta V_a}{V_a} \right)^2 +
    \left( \frac{\Delta V}{V} \right)^2
}
\end{align}

For $\frac{\Delta V_a}{V_a}$ we assume a pipette volume accuracy of 1%, to find:

````python
Ct = 1.8
dvava = 0.01
dcc = math.sqrt(0.01**2 + (dVV)**2)
print(f'dCt/Ct: {dcc:.3}')
print(f'   dCt: {dcc * Ct:.3} mM')
````

### Conclusion

If weight and total volume are measured carefully, purity of the compound is the dominant factor in the target concentration uncertainty, and we can equate a purity of 99% with a 1% error.

### Liquid junction potential

For the LJP calculation, changes to ions with low mobility have the greatest effect.
This makes it relatively robust, as the simple ions all have high mobility.

Changing Na, K, EGTA(2-) and ATP(2-) we can obtain an LJP of 6.86 mV, giving us a bracket of $6.7\pm0.2\,\text{mV}$.
