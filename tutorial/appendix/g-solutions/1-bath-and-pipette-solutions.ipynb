# G1: Bath & pipette solutions

In this appendix we look at bath and pipette solutions _from a modelling point of view_, and discuss

- The general method by which solutions are prepared
- How to determine **_free_** ion concentrations from the listed _added_ concentrations.
- How to estimate the **liquid junction potential (LJP)** and the **reversal (Nernst) potential** for potassium (EK).
- How to estimate **errors** in the free concentrations, and propagate these to LJP and EK.
- How to calculate the osmolarity of both solutions.

We assume some mathematical knowledge, but very limited practical and theoretical knowledge of chemistry.

This notebook introduces the concepts.
The next two notebooks provide real-world examples.

## General method of preparing solutions

Bath and pipette solutions are prepared by mixing powders (e.g. of [KCl](https://www.sigmaaldrich.com/GB/en/product/sigma/p9541)) and stock solutions (e.g. of [$\text{CaCl}_2$](https://www.sigmaaldrich.com/GB/en/product/sigma/21115)), adding purified water until the desired volume is reached, and adjusting the pH to e.g. 7.4 (external) or 7.2 (internal) by adding a few drops of a basic or acidic solution.

As the pH adjustment slightly increases the total volume, a more accurate method is to

1. Combine the ingredients.
2. Add pure water till just below the desired volume, and stir.
3. Adjust the pH.
4. Set the final volume.

### Adding by weight and by volume

Substances added as powder will list the molecular weight $M$ on the bottle, from which we can work out how much to add.
For example, to make $100\,\text{mL}$ solution with target concentration $10\,\text{mM}$ KCl, we would look up the molecular weight of KCl to find $74.55\,\text{g}/\text{mol}$, and then measure out
\begin{align}
0.1\,\text{L} \cdot 10\,\frac{\text{mmol}}{\text{L}} \cdot 74.55\,\frac{\text{g}}{\text{mol}} = 74.55\,\text{mg}
\end{align}
using very accurate scales.

Similarly, if we want 2 mM $\text{CaCl}_2$ in the same mix we look up the concentration, in this example $1\,\text{M}$, and use a pipette to add 
\begin{align}
0.1\,\text{L} \cdot 2\,\frac{\text{mmol}}{\text{L}} / 1\,\frac{\text{mol}}{\text{L}} = 0.2\,\text{mL}
\end{align}
of the stock solution.

### Purity

For compounds added by weight, the supplier usually lists a _purity_.
This specifies how much of the powder is actually the compound of interest.
**Although given as a lower bound, it is common practice to interpret it as the true value**, and adjust the added weight accordingly:
\begin{align}
m = \frac{C_t M V}{p} 
&\quad&
\frac{\text{mmol}}{\text{L}} \frac{\text{g}}{\text{mol}} \text{L} = \text{mg}
\end{align}
where $m$ is the weight required to obtain the target solution and $V$ is the volume of solution being prepared.
Common units are shown on the right.

For compounds added by volume, purity is often not supplied, so we simply use
\begin{align}
V_a = \frac{C_t}{C_a} V
&\quad&
\text{mL} = \frac{\text{mmol}}{\text{L}} \cdot \frac{\text{L}}{\text{mol}} \cdot \text{L}
\end{align}
where $V_a$ is the volume to add.

### "Loss on drying"

Another impurity to look out for is _water content_, for example in [this EGTA product](https://www.sigmaaldrich.com/GB/en/product/sigma/e3889).
This occurs when $\text{H}_2\text{O}$ molecules are embedded in a crystaline structure, and is measured as (but not exactly the same as) _loss on drying_. or LOD.

LOD is specified as an upper bound, for example $\leq 1\%$, but as with purity, it is commonly factored in by reinterpreting as the exact value:
\begin{align}
m = \frac{C_t M V}{p_1 (1 - p_2)} 
\end{align}
where $p_1$ is purity (e.g. 0.97 for $\geq 97\%$) and $p_2$ is loss on drying (e.g. 0.01 for $\leq 1\%$).


### Unspecified water content

When water content is significant, it is often explicitly acknowledged in the product formula.
For example, this [ATP disodium salt hydrate](https://www.sigmaaldrich.com/GB/en/product/aldrich/a26209) is listed as

\text{C}_{10}\text{H}_{14}\text{N}_5\text{Na}_2\text{O}_{13}\text{P}_3 \cdot x\text{H}_2\text{O}
\end{align}
where the $A \cdot x B$ notation means "a crystal of A with embedded B, at a ratio of $x$ B per A", see for example the wikipedia page on [hydrates](https://en.wikipedia.org/wiki/Hydrate) and [water of crystallisation](https://en.wikipedia.org/wiki/Water_of_crystallization).
By comparision, the standard formula for ATP is
\begin{align}
\text{C}_{10}\text{H}_{16}\text{N}_5\text{O}_{13}\text{P}_3
\end{align}
so in addition to the added water, we can see two Na have replaced two H.

But what is the value of $x$?
Unfortunately, the page specifies

> Contains solvent of crystallization, varying amounts of water

If we really want to know, we can look up the lot number and request the certificate of analysis from the supplier.
But for an estimate we can also look at [a different supplier](https://www.thermofisher.com/order/catalog/product/L14522.06), who list the water content as "$\leq 10\%$".
This would mean the weight to add is
\begin{align}
m = \frac{C_t M V}{p_1 (1 - p_2)} \approx 1.13 C_t M V
\end{align}
where $p_1$ is purity (0.98 for this product) and $p_2$ is water content (the estimated 0.1).

### Unspecified ratio of components

Another unspecified ratio occurs with the "$\text{Mg}^{2+}\text{ATP}$" listed in our example above.
On [the supplier website](https://www.sigmaaldrich.com/GB/en/product/sigma/a9187) the formula is given as
\begin{align}
\text{C}_{10}\text{H}_{16}\text{N}_5\text{O}_{13}\text{P}_3 \cdot x\text{Mg}^{2+}
\end{align}
which shows that this is an ATP crystal embedding magnesium.
Near the bottom of the page, the ratio of moles of Mg to moles of ATP is specified as approximately 1 to 1.5.

Note that, since we are interested in both Mg _and_ ATP, we cannot correct for both simultaneously.
Instead, we will treat this as a source of _uncertainty_.

On the product website above, the molecular weight is given as 507.18 g/mol, which is the weight of ATP.
The atomic weight of a magnesium ion is 24.304 g/mol In the worst case scenario, when we have 1.5 moles of Mg per ATP, this gives a molecular weight of 543.636 g/mol.
That means that, in addition to the purity error, measuring using the given weight will give an ATP content that is up to 7% lower than intended.
In this case, the Mg content will be 40% higher than intended.

\begin{align}
\frac{507.18}{507.18 + 1.5 \cdot 24.304} &\approx 0.933 \rightarrow 7\% \text{lower}
\end{align}
\begin{align}
0.933 \cdot 1.5 &= 1.3995 \rightarrow 40\% \text{higher}
\end{align}

## Adjusting the pH

Once we have worked out how much to add, we mix everything together and then measure the pH.
Solutions used in cardiac cell electrophysiology are typically acidic before adjustment.
For example, around 40% of HEPES molecules will release a proton ($H^+$) when dissolved, and become $\text{HEPES}^-$.
The EGTA and ATP used in internal solutions can release _multiple_ protons when dissolved.
As a result, a strong base needs to be added, and common choices are NaOH for external, and KOH for internal solutions.
**This can add a substantial amount of Na or K, particularly for internal solutions, and so will affect the LJP and reversal potentials.**

When setting up new experiments, if the exact concentrations are important it is a good idea to use a measuring pipette to add the base one drop at a time, count the drops, and calculate the added Na or K.

When analysing existing data, this amount is almost never reported.
Instead, it may be possible to estimate the added base by working out the free ion concentrations, summing all charges, and equating the deviation from zero with the amount of added base (see the **FIRST? SECOND?** example notebook).

### Do we need to keep track of $[\text{H}^+]$ and $[\text{OH}^-]$?

Since our goal is to calculate the concentrations of free ions, do we need to account for released $\text{H}^+$ and added $\text{OH}^-$ when calculating the LJP?
The short answer is **no: at a physiological pH, the $\text{H}^+$ and $\text{OH}^-$ concentrations are negligible**, compared to the other ions.
A little more background is given below.

In pure water, at pH=7, a tiny fraction of the $\text{H}_2\text{O}$ molecules ($10^{-7}\,\text{mol}/\text{L}$) [self-ionise](https://en.wikipedia.org/wiki/Self-ionization_of_water) to become $\text{H}_3\text{O}^+$ and $\text{OH}^-$.
When we add an acid such as HCl it increases the $\text{H}_3\text{O}^+$ concentration: $\text{HCl} + \text{H}_2\text{O} \rightarrow \text{H}_3\text{O}^+ + \text{Cl}^-$.
A common shorthand for this reaction is _to omit the water_, and write $\text{HCl} \rightarrow \text{H}^+ + \text{Cl}^-$.
In reality a free $\text{H}^+$ would react almost instantaneously with the nearest water molecule.
Nevertheless, we will use the "$[\text{H}^+]$" notation below.

Because in _strongly_ acidic or basic solutions, $[\text{H}^+]$ and $[\text{OH}^-]$ can vary by orders of magnitude, logarithmic measures are used:
\begin{align}
\text{pH} \approx -\log_{10} [\text{H}^+] &\quad&
\text{pOH} \approx -\log_{10} [\text{OH}^-]
\end{align}
Like in the Nernst equation, the real definitions are in terms of [activities](https://en.wikipedia.org/wiki/Thermodynamic_activity) but these are usually approximated by the concentrations.
At room temperature the two measures are related by $\text{pH} + \text{pOH} \approx 14$.

Filling in the numbers, at a pH of 7.4 we find $[\text{H}^+] \approx 4 \cdot 10^{-8}\,\text{M} = 4 \cdot 10^{-5}\,\text{mM}$, and at a pH of 7.2 we obtain $6 \cdot 10^{-5}$ mM.
Using $\text{pOH} = 14 - \text{pH}$ we can see that $[\text{OH}^-]$ is similarly small.
**As a result, $[\text{H}^+]$ and $[\text{OH}^-]$ can be neglected in LJP calculations at physiological pH**.
Similarly, they can be omitted when calculating the total charge of the solution, or its "ionic equivalence" (see below).

## From target concentration to free concentrations

So far we have concentrated on the target concentrations.
The next question is what happens once all these ingredients dissolve and mix.
Salts like NaCl or KCl dissolve practically completely in water, and since there are typically no substances binding Na, K, or Cl in our solutions, we can assume that adding 10mM NaCl adds 10mM free $[\text{Na}^+]$ and 10mM free $[\text{Cl}^-]$.

But other compounds are more complicated, including

1. Buffers of the pH such as HEPES, which will exist in a protonated fraction ($[\text{HEPES}]$) and a deprotonated fraction ($[\text{HEPES}^-]$) which depends on the pH.
2. Other compounds with multiple protonation states, such as EGTA, which is added as uncharged $\text{H}_4\text{EGTA}$, but can lose up to four protons to become $\text{EGTA}^{4-}$. At physiological pH, it exists almost exclusively as $\text{H}_2\text{EGTA}^{2-}$.
3. Compounds that bind to ions, for example ["chelation"](https://en.wikipedia.org/wiki/Chelation) of $\text{Ca}^{2+}$ by EGTA, and of $\text{Mg}^{2+}$ by ATP.

## Buffering with HEPES

HEPES can exist as either $[\text{HEPES}]$ or $[\text{HEPES}^-]$, with the exact ratio between the two depending on the pH.
This is described by the [Henderson-Hasselbalch equation](https://en.wikipedia.org/wiki/Henderson%E2%80%93Hasselbalch_equation):
\begin{align}
\text{pH} = \text{pK}_a + \log_{10}\left( \frac{[\text{HEPES}^-]}{[\text{HEPES}]} \right)
\end{align}
Where $\text{pK}_a = -\log_{10} K_a$ is HEPES's [acid dissociation rate constant](https://en.wikipedia.org/wiki/Acid_dissociation_constant).
The "a" in "pKa" stands for "acid", which can be confusing, as the notation $K_a$ is also used to mean a general _association_ constant (and so equal to $1/K_a$ in the "a for acid disassociation" notation).

We can combine the equation above with
\begin{align}
[\text{HEPES}^-] + [\text{HEPES}] = [\text{HEPES}]_\text{added}
\end{align}
to find
\begin{align}
[\text{HEPES}] = \frac{[\text{HEPES}]_\text{added}}{1 + 10^{\text{pH} - \text{pK}_a}}
&&\quad&&
[\text{HEPES}^-] = \frac{[\text{HEPES}]_\text{added}}{1 + 10^{\text{pK}_a - \text{pH}}}
\end{align}

A value for HEPES' pKA can be found e.g. [on this website](https://condor.depaul.edu/jmaresh/341/Lab1/Lab_1_Buffers4.html), which lists it as 7.55 at $20\,^\circ\text{C}$, with a temperature dependent shift of -0.014 per degree celsius.
So, for example, if the pH of a solution with 10mM HEPES was set to 7.4 at $22\,^\circ\text{C}$, the pKa was 7.522, yielding
\begin{align}
[\text{HEPES}] = 5.7\,\text{mM}
&&\quad&&
[\text{HEPES}^-] = 4.3\,\text{mM}
\end{align}

### Temperature change

What happens if the temperature is changed _after the pH was set_?

Here, [the same website](https://condor.depaul.edu/jmaresh/341/Lab1/Lab_1_Buffers4.html) says something confusing:

> ... if you calculate the change in pKa with temperature, the pH of the buffer will change the same amount because changing the temperature will not change the ratio of the basic form to acid form ... raising the temperature of HEPES by 1.0 °C lowers both the pKa and pH by 0.014.

In the light of the equation above, this sounds impossible: for the pH to change, HEPES must lose or gain protons, and so the protonation ratio _must_ change.
However, changing the pH from e.g. 7.2 to 7.1 implies a change from $6.3 \cdot 10^{-5}\,\text{mM}$ free $\text{H}^+$ to $7.9 \cdot 10^-5\,\text{mM}$, so that the HEPES protonation ratio is _practically_ unchanged.

If we set the pH at 22 degrees, but measure at $37\,^\circ\text{C}$, the ratio of charged to uncharged HEPES will be roughly the same, and we can ignore the temperature change in our calculations.
The pH, however, will have changed.
**For this reason, it is always good to either set the pH at the temperature the experiments will be run**, or alternatively to calculate the expected shift and adjust for this accordingly.

## $\text{Ca}^{2+}$, EGTA, $\text{Mg}^{2+}$, and ATP

The free calcium concentration inside a cell is very low, and tightly controlled.
To recreate this situation in pipette solution, it is common to add a very strong calcium buffer, such as EGTA, and to choose the buffer-to-calcium ratio to achieve a specific micromolar free calcium concentration.

ATP, the main source of energe inside cells, [requires magnesium to be active](https://en.wikipedia.org/wiki/Magnesium_in_biology)) and is present inside cells mostly as $\text{MgATP}^{2-}$.
To recreate this in pipette solution, we either add both seperately (e.g. with $\text{MgCl}_2$ and [ATP disodium salt](https://www.sigmaaldrich.com/GB/en/product/aldrich/a26209)) or already together (e.g. with [ATP magnesium salt hydrate](https://www.sigmaaldrich.com/GB/en/product/sigma/a9187)).

In addition, EGTA can bind magnesium, and ATP can bind calcium.
And so, **if accurate estimates of free calcium are required**, all four must be considered simultaneously.
However, **approximate methods considering Ca-EGTA and Mg-ATP in isolation are sufficient to calculate the LJP**.

### Calcium and EGTA in isolation

EGTA is a strong and highly specific calcium buffer.
In a solution with more EGTA than calcium, almost all calcium will be bound, and a change to the _total_ amount of calcium, $[\text{Ca}]_\text{total} = [\text{Ca}^{2+}]_\text{free} + [\text{Ca}]_\text{bound}$, will lead to a much smaller change in the amount of _free_ calcium.

Besides its effect on $[\text{Ca}^{2+}]_\text{free}$, we need to take EGTA into account for the LJP because both free EGTA and bound CaEGTA can exist in multiple protonation states, and therefore have charges (see [Dweck, Reyes-Alfonso Jr., and Potter](https://doi.org/10.1016/j.ab.2005.09.025)).
To distinguish between the protonation states, we will refer to the [canonical molecule](https://en.wikipedia.org/wiki/EGTA) $\text{C}_{14}\text{H}_{24}\text{N}_2\text{O}_{10}$ as "$\text{H}_4\text{EGTA}$", and to the common deprotonated forms as $\text{H}_3\text{EGTA}^{-}$, $\text{H}_2\text{EGTA}^{2-}$ etc.

To estimate the concentrations of free and bound calcium and EGTA (at physiological pH and room or body temperature), we make the following simplifying assumptions:

1. All EGTA and $\text{Ca}^{2+}$ that can bind, will bind.
2. All bound EGTA is present as $\text{Ca}\text{EGTA}^{2-}$
3. All free EGTA is present as $\text{H}_2\text{EGTA}^{2-}$


Under these assumptions, our example solution with target concentrations of $1\,\text{mM}\,[\text{Ca}^{2+}]$ and $10\,\text{mM}\,\text{EGTA}$ will have $[\text{Ca}^{2+}]_\text{free} = 0\,\text{mM}$, $[\text{CaEGTA}^{2-}] = 1\,\text{mM}$, and $[\text{H}_2\text{EGTA}^{2-}] = 9\,\text{mM}$.


### Magnesium and ATP in isolation

Just like EGTA, ATP can lose up to four protons, and we start by introducing the notation "$\text{H}_4\text{ATP}$" for its [canonical form](https://en.wikipedia.org/wiki/Adenosine_triphosphate) $\text{C}_{10}\text{H}_{16}\text{N}_5\text{O}_{13}\text{P}_3$, and then make the following simplifying assumptions:

1. All ATP and $\text{Mg}^{2+}$ that can bind, will bind.
2. All bound ATP is present as $\text{MgATP}^{2-}$.
3. All free ATP is present as either $\text{ATP}^{4-}$ or $\text{HATP}^{3-}$.

In our example solution with target concentrations $[\text{Mg}^{2+}]= 6\,\text{mM}$ and $[\text{ATP}]= 5\,\text{mM}$ this gives us $[\text{Mg}^{2+}]=1\,\text{mM}$, $[\text{MgATP}^{2-}]=5\,\text{mM}$, and $[\text{ATP}]_\text{free}=0\,\text{mM}$.

In situations where we _do_ have free ATP, we can split $[\text{ATP}]_\text{free}$ into $[\text{ATP}^{4-}]$ and $[\text{HATP}^{3-}]$ using the same equations as for HEPES, and the Pka value of 6.5 given by [Stockbridge & Wolfenden](https://doi.org/10.1074/jbc.M109.017806)).
\begin{align}
[\text{ATP}^{4-}] = \frac{[\text{ATP}]_\text{free}}{1 + 10^{\text{pK}_a - \text{pH}}}
&&\quad&&
[\text{HATP}^{3-}] = \frac{[\text{ATP}]_\text{free}}{1 + 10^{\text{pH} - \text{pK}_a}}
\end{align}
where
\begin{align}
[\text{ATP}]_\text{free} = [\text{ATP}^{4-}] + [\text{HATP}^{3-}]
\end{align}

### Considering all four at once

If the free calcium concentration is critical for an experiment, interactions between $\text{Ca}^{2+}$ and ATP and between $\text{Mg}^{2+}$ and EGTA must also be considered, and this can be done using online calculators.
Most famously, [Maxchelator](https://somapp.ucdmc.ucdavis.edu/pharmacology/bers/maxchelator/downloads.htm), described in [Bers, Patton & Nuccitelli 2010](https://doi.org/10.1016/S0091-679X(08)61108-5) is still available online (although backing up the HTML and hardcoded JS is recommended).
A more recent effort is [PyChelator](https://amrutelab.github.io/PyChelator/), which is available as a Colab notebook or in Javascript, and described in [Spahiu, Kastrati & Amrute-Nayak 2024](https://doi.org/10.1186/s12859-024-05858-8), with code on [GitHub](https://github.com/AmruteLab/PyChelator).

Both programs require the following input:

- Total (added) concentration of $\text{Ca}^{2+}$, $\text{Mg}^{2+}$, EGTA, and ATP
- The temperature and pH
- The "ionic equivalence", $I_e$, of the solution

The last value is found as
\begin{align}
I_e = \frac{1}{2} \sum{C_i |z_i|}
\end{align}
where $C_i$ is the concentration of an ionic species, $|z_i|$ is the magnitude of its valence, and the sum is over all (charged) species in the solution.
For the choice of ionic equivalence over the more conventional ionic strength ($\frac{1}{2} \sum{C_i z_i^2}$), see [Smith & Miller, 1984](https://doi.org/10.1016/0304-4165(85)90011-X).

The need for an ionic equivalence presents us with a slight problem, as it depends on the ionic concentrations we need to calculate.
Fortunately, the influence of $I_e$ is not particularly strong, and so we can use an estimate, for example based on the assumptions made above.

To do the calculation, both programs
1. Load a set of binding constants from a database. Here, select the one that matches your experimental conditions best.
2. Adjust the constants based on pH, temperature, and $I_e$. As described in the Bers et al. paper above.
3. Make an initial guess, and iteratively refine it.

Both programs then return the free and bound fractions Ca, Mg, EGTA, and ATP, but unfortunately not the exact concentrations of $[\text{CaEGTA}^{2-}]$, $[\text{MgEGTA}^{2-}]$, etc.
In the Example notebooks we show the simplifying assumptions above can be used to work these out.

## Calculating the LJP and reversal potentials

Using the above, we can make a table of all free ionic species in the pipette and bath solutions.
These can then be fed into LJP calculators such as [LJPCalc](https://swharden.com/LJPcalc/).

Reversal potentials for single-ion currents are calculated from the [Nernst equation](https://en.wikipedia.org/wiki/Nernst_equation):
\begin{align}
E = \frac{RT}{F} \log \frac{[\text{X}]_e}{[\text{X}]_i}
\end{align}
where the subscripts $e$ and $i$ denote external (bath) and internal (pipette).
For multi-ion currents, the [Goldman–Hodgkin–Katz voltage equation](https://en.wikipedia.org/wiki/Goldman_equation) can be used, but note that, for the canonical example - the L-type calcium current - we do need an accurate estimate of the free calcium concentration.

## Estimating the uncertainty in target concentrations

Next, we show how to estimate the uncertainty in calculated target concentrations due to uncertainty in total volume, added weight or volume, and purity of the added compounds.
To do this, we use the method outlined below.

### Propagation of uncertainty

Given any function $f$ of a measurement $x \pm \Delta x$, we can estimate the error in $f(x)$ by taking a partial derivative and linearising:
\begin{align}
\Delta f = \left| \frac{\partial f}{\partial x} \right| \Delta x
\end{align}
For multiple _independent_ measurements, we can extend this to e.g.
\begin{align}
f(x, y, z) \rightarrow \Delta f = \sqrt{
    \left( \frac{\partial f}{\partial x} \Delta x \right)^2 +
    \left( \frac{\partial f}{\partial y} \Delta y \right)^2 +
    \left( \frac{\partial f}{\partial z} \Delta z \right)^2}
\end{align}
See e.g. [wikipedia on "propagation of uncertainty"](https://en.wikipedia.org/wiki/Propagation_of_uncertainty#Non-linear_combinations).
For intuition, we can use probability theory to interpret $\Delta f$ as a variance, or think geometrically and interpret it as a distance.

Below, we write the target concentration as a function of the added weight/volume, the final volume, and purity, and work out how their measurement accuracy affects the final concentration.

### Adding by weight

To estimate the error when adding substances by weight, we start by writing the equation for target concentration $C_t$ as a function of total volume $V$, added weight $m$, molecular weight $M$, and purity $p$:
\begin{align}
C_t = \frac{m \cdot p}{M \cdot V} && \quad && \frac{\text{mmol}}{\text{L}} = \frac{\text{mg}}{\text{g}/\text{mol} \cdot \text{L}}
\end{align}

Here, $V$ and $m$ depend on the accuracy of the measuring equipment ($\Delta V$ and $\Delta m$), so we will have at least two terms in our error equation:
\begin{align}
\frac{\partial C_t}{\partial m} \Delta m &= \frac{f}{M \cdot V} \Delta m = C_t\frac{\Delta m}{m} \\
\frac{\partial C_t}{\partial V} \Delta V &= \frac{-m \cdot f}{M \cdot V^2} \Delta V = - C_t \frac{\Delta V}{V}
\end{align}

Next, purity, water content, and unknown ratios can all be captured in the variable $p$.
But to do this, we need to decide how to interpret the lower bounds given by manufacturers.
For example, a purity of 99% can optimistically be interpreted as a range $[0.99, 1]$ for $p = 0.995$ and $\Delta p = 0.005$, giving an extra term

\begin{align}
\frac{\partial C_t}{\partial p} \Delta p &= \frac{m}{M \cdot V} \Delta p = C_t\frac{\Delta f}{p} \\
\end{align}

Combining, and separating the repeated $C_t$ term, we find
\begin{align}
\Delta C_t &= C_t\sqrt{\left(\frac{\Delta m}{m}\right)^2 + \left(\frac{\Delta V}{V}\right)^2 + \left(\frac{\Delta p}{p}\right)^2}
\end{align}
This means that we can do the analysis either in terms of absolute errors (e.g. $\frac{f}{M \cdot V} \Delta m$) or relative errors (e.g. $\frac{\Delta m}{m}$).

### Adding by volume

Similarly, for the target concentration when adding a volume $V_a$ of stock solution with molar concentration $C_a$, we find
\begin{align}
C_t = \frac{C_a \cdot V_a}{V}
&& \quad &&
\frac{\text{mmol}}{\text{L}} = \frac{\text{mmol}}{\text{L}} \frac{\text{mL}}{\text{mL}}
\end{align}
and
\begin{align}
\Delta C_t = C_t \sqrt{
    \left( \frac{\Delta V_a}{V_a} \right)^2 +
    \left( \frac{\Delta V}{V} \right)^2 +
    \left( \frac{\Delta C_a}{C_a} \right)^2
}
\end{align}
where the final term can be omitted in the common scenario that no uncertainty bound in $C_a$ was provided.

## Osmolarity

Finally, we very briefly discuss [osmolarity](https://en.wikipedia.org/wiki/Osmotic_concentration).
This is very important when _devising a set of solutions_, as a mismatch between solutions will cause cells to shrink or swell.
Although in patch clamp, it is seen as [good practice](https://www.scientifica.uk.com/neurowire/electrophysiology-what-goes-on-the-inside) to give the external solution a slightly higher osmolarity.

Osmolarity is expressed in _Osmoles_ (Osm), with one osmole [defined as](https://en.wiktionary.org/wiki/osmole):

> [one] mole of an osmotically active compound

so if we decide all our compounds are "osmotically active", we can equate 1 mM with 1 mOsm.

Osmolarity can be estimated directly from the target concentrations, using
\begin{align}
\sum \phi_i n_i C_{t,i} \approx \sum n_i C_{t,i}
\end{align}
where $C_{t, i}$ is the target concentration for compound $i$ and $n_i$ is the number of particles the compound dissociates into (e.g. 2 for NaCl).
The [osmotic coefficient](https://en.wikipedia.org/wiki/Osmotic_coefficient) $\phi_i \approx 1$ can be used to account for non-ideality of the dissociation.

Typical values for $n_i$ are 1 for HEPES, glucose, and sucrose; 2 for NaCl and KCl; and 3 for CaCl2.
But for Ca, Mg, EGTA, and ATP we would need to take binding after disassociation into account, and arrive at fractional $n_i$.
Instead, we can use our estimated _free concentrations_ to calculate
\begin{align}
\sum C_{f,i}
\end{align}
