**This example** describes a set of solutions used in experiments on multiple ion channels.
They were prepared and pH adjusted at a room temperature of $20.5,^\circ\text{C}$, and patch clamp experiments were performed at the same temperature.

**Summary**: In this example, the compounds are all very pure, but total volume is set with a 2% precision, which becomes the main source of error.

### Charge balance and the pH

The final charge balance is off by 1.4 mM.
Since NaCl, KCl, MgCl2 and CaCl2 are all electrically neutral and do not react with water, they cannot contribute to this error, which instead must be due to:

- the NaOH measurement or stock concentrations (mL added, concentration of added solution, total volume)
- the HEPES concentration (weight, total volume, pH)

## Uncertainty in the target concentrations

Finally, we look at uncertainty in the target concentrations that can arise when measuring out ingredients.

### Adding by weight

To estimate the uncertainty in concentration for substances added by weight, we need to assign values to the terms in
\begin{align}
\Delta C_t &= C_t\sqrt{\left(\frac{\Delta m}{m}\right)^2 + \left(\frac{\Delta V}{V}\right)^2 + \left(\frac{\Delta p}{p}\right)^2}
\end{align}

In this experiment, the external volume was set using a 1L chemical beaker, with an accuracy of 20mL, giving
\begin{align}
\frac{\Delta V}{V} = \frac{20}{1000} = 0.02
\end{align}

As in Example 1, we will use $\delta_m = 1\,\text{mg}$.

Filling in these values for NaCl, KCl and HEPES, we find:

````python
# Ct, M, purity
ions = {
    'NaCl': (140, 58.44, 99.5),
    'KCl': (5.4, 74.55, 99),
    'HEPES': (15, 238.3, 99.5),
}

# Relative error in final volume
dVV = 0.02

# Absolute error in weight
dm = 1

for k, (Ct, M, p) in ions.items():
    p = p / 100
    dp = 1 - p
    m = Ct * M * 1 / p
    print(f'{k} (Ct={Ct} mM, m={m:.0f} mg)')
    dcc = math.sqrt(dVV**2 + (dp / p)**2 + (dm / m)**2)
    print(f'  |dV/V|: {abs(dVV):1.6g}')
    print(f'  |dp/p|: {abs(dp/p):1.6g}')
    print(f'  |dm/m|: {abs(dm / m):1.6g}')
    print(f'  dCt/Ct: {dcc:1.6f}')
    print(f'     dCt: {dcc * Ct:1.6f} mM')
````

So in this experiment, the relative error due to the final volume dominates, and we can use 2% as an estimate for all external compounds.
Similarly, the internal volume was set to 100mL with a 2mL error, for another 2% error.

Both errors are larger than the volume increase from pH adjustment, which we will therefore ignore.

The lowest purity in this experiment was for EGTA, which we will treat below:

````python
# Ct, M, purity
ions = {
    'EGTA': (5, 380.35, 97),
}

# Relative error in final volume
dVV = 0.02

# Absolute error in weight
dm = 1

for k, (Ct, M, p) in ions.items():
    p = p / 100
    dp = 1 - p
    m = Ct * M * 1 / p
    print(f'{k} (Ct={Ct} mM, m={m:.0f} mg)')
    dcc = math.sqrt(dVV**2 + (dp / p)**2 + (dm / m)**2)
    print(f'  |dV/V|: {abs(dVV):1.6g}')
    print(f'  |dp/p|: {abs(dp/p):1.6g}')
    print(f'  |dm/m|: {abs(dm / m):1.6g}')
    print(f'  dCt/Ct: {dcc:1.6f}')
    print(f'     dCt: {dcc * Ct:1.6f} mM')
````

Note that, since the 2% errors are all due to the total volume these errors are not independent.
Instead, the relative error must be the same for all internal concentrations, and for all external concentrations.

For external sodium and internal potassium, we could also have included the final charge imbalance (assuming that the added NaOH or KOH was measured inaccuratly), but this would not significantly increase the combined error.

For the external error of 1.4 mM:
\begin{align}
\sqrt{3^2 + 1.4^2} \approx 3.31
\end{align}
