# G2: Example 2, multi-channel experiments

**This example** describes a set of solutions used in experiments on multiple ion channels.
They were prepared and pH adjusted at a room temperature of $20.5,^\circ\text{C}$, and patch clamp experiments were performed at the same temperature.

**Summary**: In this example, the compounds are mostly very pure but total volume is set with a 2% precision, which becomes the main source of error.
We briefly discuss the accuracy of scales and the people using them (in this case the author).

## Uncertainty in the target concentrations

Finally, we look at uncertainty in the target concentrations that arise when measuring out ingredients.

### Adding by weight

To estimate the uncertainty in concentration for substances added by weight, we need to assign values to the terms in
\begin{align}
\frac{\Delta}{C} &= \sqrt{\left(\frac{\Delta m_a}{m_a}\right)^2 + \left(\frac{\Delta V}{V}\right)^2 + \left(\frac{\Delta q}{q}\right)^2}
\end{align}

#### $m_a$ and $\Delta m_a$

In this experiment, the added weights $m_a$ were recorded.
The table below lists the intended weights, $m$, in addition to actually added weight (read directly from the scales), for the external solution:

| Compound | $C_t$    | $M$          | $p$           | $m$       | $m_a$   |
|----------|----------|--------------|---------------|-----------|---------|
| NaCl     | 140 mM   |  58.44 g/mol | $\geq 99.5\%$ | 8181.6 mg | 8178 mg |
| KCl      |   5.4 mM |  74.55 g/mol | $\geq 99\%$   |  402.6 mg |  402 mg |
| HEPES    |  15 mM   | 238.3  g/mol | $\geq 99.5\%$ | 3574.5 mg | 3570 mg |

The internal solution was made more carefully:

| Compound    | $C_t$    | $M$          | $p$           | $m$       | $m_a$     |
|-------------|----------|--------------|---------------|-----------|-----------|
| KCl         |  10 mM   |  74.55 g/mol | $\geq 99\%$   |   74.6 mg |   74.4 mg |
| K-Gluconate | 125 mM   | 234.25 g/mol | $\geq 99\%$   | 2928.1 mg | 2929.0 mg |
| EGTA        |   5 mM   | 380.35 g/mol | $\geq 97\%$   |  190.2 mg |  190.7 mg |
| Na2ATP      |   4 mM   | 551.1 g/mol  | $\geq 99\%$ * |  220.4 mg |  220.6 mg |
| HEPES       |   5 mM   | 238.3 g/mol  | $\geq 99.5\%$ |  119.2 mg |  119.5 mg |

For $\Delta m_a$ we can look at specifications of "analytical balances", for example [this one](https://www.mt.com/us/en/home/phased_out_products/Laboratory_Weighing_Solutions/Analytical/Excellence/XS_Analytical_Balance/XS64.html) used in Example 2.
It shows several measures, including readability (how many digits are on the display, regardless of accuracy), repeatability (the standard deviation obtained if you keep measuring the same thing), and linearity (if you measure objects over the full allowable range, what is the maximum deviation from the true weight).
To find an upper bound, we could use the largest listed number of $\pm0.2\,\text{mg}$.
However, this precision is only achieved if the scales are used with extreme care, so that something larger, for example $\pm1\,\text{mg}$ may be more realistic.

In addition to precision of the _scales_, we can factor precision of the experimenter into $\Delta m_a$.
Above, the experimenter accepted errors of 4, 0.6, and 4.5 mg in the external solution, and, on a different day, errors of 0.2, 0.9, 0.5, 0.2, and 0.3 mg.
We can use these numbers to set an approximate $\Delta m_a$ in cases where $m_a$ is unknown, using $\Delta m_a = 1\,\text{mg}$ for a precisely made solution (so that the error of the scales dominates), and $\Delta m_a = 5\,\text{mg}$ in a less precisely made solution.

#### $\Delta V / V$

In this experiment, the external volume was set using a 1L chemical beaker, with an accuracy of 20mL, giving
\begin{align}
\frac{\Delta V}{V} = \frac{20}{1000} = 0.02
\end{align}

#### Purity

Because purity was not taken into account, we will incur both a bias and a relative error due to $p$.

We will factor in purity using 
\begin{align}
C = \frac{m_a \cdot q}{M \cdot V}
& \quad &
q = \frac{1 + p}{2}
& \quad &
\Delta q = \frac{1 - p}{2}
& \quad &
\frac{\Delta q}{q} = \frac{1 - p}{1 + p}
\end{align}

### External solution

Filling in for NaCl, KCl and HEPES, we find:

````python
# Ct, M, purity, ma
ions = {
    'NaCl': (140, 58.44, 99.5, 8178),
    'KCl': (5.4, 74.55, 99, 402),
    'HEPES': (15, 238.3, 99.5, 3570),
}

# Volume and relative error in final volume
V = 1
dVV = 0.02

# Absolute error in weight
dm = 1

for k, (Ct, M, p, ma) in ions.items():
    p = p / 100
    q = (1 + p) / 2
    dq = (1 - p) / 2
    C = ma * q / (M * V)
    print(f'{k} (Ct={Ct} mM)')
    dcc = math.sqrt(dVV**2 + (dm / ma)**2 + (dq / q)**2)
    dc = dcc * C
    print(f'  |dV/V|: {abs(dVV):1.6g}')
    print(f'  |dm/m|: {abs(dm / ma):1.6g}')
    print(f'  |dq/q|: {abs(dq / q):1.6g}')
    print(f'   dC/C : {dcc:1.6f}')
    print(f'   dC   : {dc:1.6f} mM')
    print(f'    C   : {C:6.2f} [{C - dc:6.2f}, {C + dc:5.1f}]')
````

This shows that, in this example:

1. The uncertainty in total volume dominates, and the relative errors in NaCl, KCl, and HEPES concentration are all 2 to 2.1%.
2. Because of the high purities used, the bias due to not correcting for purity is low.
3. The uncertainty in added weight is is negligible, even for KCl which had the smallest added weight

The volume increase from pH adjustment was 1.15 mL, and not included in the analysis above.
But as this was a 0.1% increase, it will be dwarfed by the 2% error and so is safely ignored.

### Adding by volume

The substances added by volume were added with a precision pipette, and so here as well the 2% volume error will dominate.

### Internal solution

To repeat this for the internal solution, we need to deal with the unspecified water content in $\text{Na}_2\text{ATP}$.
Although we do not have the lot number for this data, we can look at the 5 last published certificates of analysis (using the product link above).
Here, we see that

1. The upper limit is 12% water
2. The last five water lots have water content 7.1%, 6.4%, 6.4%, 9.0%, 4.7%.

So it looks like values are spread out over the interval, not clustered near the upper limit.
An easy way to incorporate the water content then, is to ignore the higher purity (99%), and fill in $q=0.88$ instead.

We then find:

````python
# Ct, M, purity, ma
ions = {
    'KCl': (10, 74.55, 99, 74.4),
    'K-Gluconate': (125, 234.25, 99, 2929),
    'EGTA': (5, 380.35, 96, 190.7),  # Combined purity and lod
    'Na2ATP': (4, 551.1, 88, 220.6),
    'HEPES': (5, 238.3, 99.5, 119.5),
}

# Volume and relative error in final volume
V = 1
dVV = 0.02

# Absolute error in weight
dm = 1

for k, (Ct, M, p, ma) in ions.items():
    p = p / 100
    q = (1 + p) / 2
    dq = (1 - p) / 2
    C = ma * q / (M * V) / 0.1
    print(f'{k} (Ct={Ct} mM)')
    dcc = math.sqrt(dVV**2 + (dm / ma)**2 + (dq / q)**2)
    dc = dcc * C
    print(f'  |dV/V|: {abs(dVV):1.6g}')
    print(f'  |dm/m|: {abs(dm / ma):1.6g}')
    print(f'  |dq/q|: {abs(dq / q):1.6g}')
    print(f'   dC/C : {dcc:1.6f}')
    print(f'   dC   : {dc:1.6f} mM')
    print(f'    C   : {C:6.2f} [{C - dc:6.2f}, {C + dc:5.1f}]')
````
