**This example** describes a set of solutions used in $\text{I}_\text{Kr}$ experiments (and as the example in the previous notebook).
They were prepared and pH adjusted at a room temperature of $22\pm1\,^\circ\text{C}$, but patch clamp experiments were performed at $37^\circ\text{C}$.

**Summary**: In this example, the total volume is set very carefully, and the purity of the compounds becomes the main source of error.
We show how PyChelator can be used to calculate Ca-Mg-EGTA-ATP interactions, but how this does not affect the LJP.
The pH adjustment shifts the reversal potential by several millivolts.

## Reversal potential

Accepting these results, we can calculate the reversal potential for potassium from
\begin{align}
E &= \frac{RT}{F} \log \frac{[\text{K}^+]_e}{[\text{K}^+]_i}
\end{align}

Using $37^\circ\text{C}$ and the values from the table, we find $E_K = -99.6\,\text{mV}$.

Neglecting the pH adjustment, we would get $E_K(130\,\text{mM}) = -93.0\,\text{mV}$.

````python
import math

T = 273.15 + 37
R = 8314  # J/kmol/K
F = 96485 # C/mol
R * T / F * math.log(4 / 166)
````

## Uncertainty in the target concentrations

Finally, we look at uncertainty in the target concentrations that can arise when measuring out ingredients.

### Adding by weight

To estimate the uncertainty in concentration for substances added by weight, we need to assign values to the terms in
\begin{align}
\frac{\Delta C}{C} &= \sqrt{\left(\frac{\Delta m_a}{m_a}\right)^2 + \left(\frac{\Delta V}{V}\right)^2 + \left(\frac{\Delta q}{q}\right)^2}
\end{align}
where $m_a$ is the added weight, $q$ is a scaling factor for purity (see the previous notebook), and $C$ is the obtained concentration
\begin{align}
C = \frac{m_a \cdot q}{M \cdot V}
\end{align}

#### $\Delta V / V$

In this experiment, the external volume was set using a [highly accurate](https://medilabexports.com/class-a-vs-class-b-laboratory-glassware/) 1 litre _volumetric flask_.
These have a tolerance of either $\pm 0.3\,\text{mL}$ or $\pm 0.6\,\text{mL}$ so we can set an upper bound for $\frac{\Delta V}{V}$ of at $6\cdot10^{-4}$.

#### $\Delta m_a / m_a$

Next, we need to calculate the relative errors $\frac{\Delta m_a}{m_a}$.
**In this example, we shall assume that the measured out weights $m_a$ are not known**.
Instead, we use the table above to calculate the smallest weight added as $298.2\,\text{mg}\,\text{KCl}$, or $301.2\,\text{mg}$ if adjusted for purity.
We combine this with an estimate of $\Delta m_a = 1\,\text{mg}$ (from the next notebook!) for an an upper limit $\frac{\Delta m_a}{m_a} \leq 3 \cdot 10^{-3}$.

#### Purity

Since the method to calculate $m_a$ is unknown, an we use $q = 1$ and $\Delta q = 1 - p$.

#### External solution

Filling in for NaCl, KCl and HEPES, we find:

````python
import math

# Ct, Purity
ions = {
    'NaCl': (140, 99),
    'KCl': (4, 99),
    'HEPES': (10, 99.5),
}

# Relative error in final volume
dVV = 6e-4

# Relative error in added weight
dmm = 3e-3

for k, (Ct, p) in ions.items():
    p = p * 0.01
    dp = 1 - p
    print(f'{k} (Ct={Ct} mM)')
    dcc = math.sqrt(dVV**2 + dmm**2 + (dp / p)**2)
    print(f'  |dV/V|: {abs(dVV):1.6g}')
    print(f'  |dm/m|: {abs(dmm):1.6g}')
    print(f'  |dp/p|: {abs(dp / p):1.6g}')
    print(f'  dCt/Ct: {dcc:1.6f}')
    print(f'     dCt: {dcc * Ct:1.6f} mM')
````

So in this experiment, the relative error in sodium, concentration, and HEPES concentration is

1. Small, being at most 1.1%
2. Dominated by $\Delta q$

For the absolute error, we see that the compound with the highest concentration has the largest error.

### Adding by volume

For $\text{CaCl}_2$, which was added by volume, we need to assign values to
\begin{align}
\frac{\Delta C}{C} = \sqrt{
    \left( \frac{\Delta V_a}{V_a} \right)^2 +
    \left( \frac{\Delta V}{V} \right)^2
}
\end{align}

For $\frac{\Delta V_a}{V_a}$ we assume a pipette volume accuracy of 1%, to find:

````python
Ct = 1.8
dvava = 0.01
dcc = math.sqrt(0.01**2 + (dVV)**2)
print(f'dCt/Ct: {dcc:.3}')
print(f'   dCt: {dcc * Ct:.3} mM')
````

### EGTA and Mg-ATP

In the interal solution, we expect the above pattern to be repeated, with weight and volume contributing no signifcant error.

For EGTA, we then have a relative error due to purity (97%) and loss on drying (1%).
Instead of treating these as separate values, we will multiply them together into a single factor
\begin{align}
p = 0.97 (1 - 0.01) \approx 0.96
\end{align}
We assume no knowledge of whether purity was included in the added weight calculation, and so use $1 - p$ as the releative error in both:

For Mg-ATP, we estimated in the previous notebook that there may be up to 7% less ATP, and up to 40% more Mg.
Because adjusting for both is impossible, here we could use the "purity not included" method from the previous notebook.
However, to keep things simpler, we will simply write out the ranges:

| Ion                  | mM   | Range        | Range $\pm 5\%$ |
|----------------------|------|--------------|-----------------|
| $\text{Mg}^{2+}$     | 1    | [1, 1.4]     | [1, 1.5]        |
| $\text{MgATP}^{2-}$  | 4.5  | [4.2, 4.5]   | [4, 4.7]        |
| $\text{MgEGTA}^{2-}$ | 0.5  | [0.5, 0.7]   | [0.5, 0.7]      |
| $\text{HATP}^{3-}$   | 0.09 | [0.08, 0.09] | [0.08, 0.09]    |
| $\text{ATP}^{4-}$    | 0.46 | [0.43, 0.46] | [0.41, 0.48]    |

To add the final complication, we have extended the ranges with the 5% purity error incurred in addition to the unknown ratio error.
We have not, however, recalculated the expected ATP protonation ratios, EGTA-Mg interaction, etc.
But, **looking at the previous results showing the limited effect of including chelation intricacies, we can safely assume the Mg to ATP ratio does not affect the LJP in this experiment**.

### Uncertainties: Conclusion

If weight and total volume are measured carefully, purity of the compound is the dominant factor in the target concentration uncertainty, and we can equate a purity of 99% with a 1% error.
The exception is Mg-ATP, which has a high relative error but is present in low concentrations.

### Liquid junction potential

For the LJP calculation, changes to ions with low mobility have the greatest effect.
This makes it relatively robust, as the ions with the highest concentrations, and therefore the largest absolute uncertainty, all have high mobility.
